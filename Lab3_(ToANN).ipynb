# 1.1 Cобрать реализацию модели нейронов с функцией активации:

- линейной
- сигмоидной
- тангенциальной
- сложной сигмоидной

### Линейный нейрон

```python
def linear_neuron(x: float = None, w: float = None, bias: float = 0):
    s = np.dot(w, x.T) + bias
    y = s
    return y
```

### Сигмоидный нейрон

```python
def sigmoid_neuron(x: float = None, w: float = None, bias: float = 0):
    s = np.dot(w, x.T) + bias
    y = 1 / (1 + np.exp(-s))
    return y
```

### Сигмоидный тангенциальный нейрон

```python
def sigmoidt_neuron(x: float = None, w: float = None, bias: float = 0):
    s = np.dot(w, x.T) + bias
    y = 2 / (1 + np.exp(-s)) - 1
    return y
```

### Сложный сигмоидный нейрон

```python
def sigmoidt_complex_neuron(x: float = None, w: float = None, bias: float = 0, lymbd: float = 1):
    s = np.dot(w, x.T) + bias
    y = 2 / (1 + np.exp(-lymbd * s)) - 1
    return y
```

# 1.2 Построить визуализацию выходов на одинаковом множестве входов

```python
x = np.linspace(-10, 10.1, 100).reshape((100, 1))
w = np.array([0.9]).reshape((1, 1))
bias = 0.3
lymbd = 2

y_linear = linear_neuron(x=x, w=w, bias=bias)

plt.figure(figsize=(10, 6))
plt.plot(x[:, 0], y_linear[0, :])
plt.title('Линейный нейрон')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

y_sigmoid = sigmoid_neuron(x=x, w=w, bias=bias)

plt.figure(figsize=(10, 6))
plt.plot(x[:, 0], y_sigmoid[0, :])
plt.title('Сигмоидный нейрон')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

y_tanh = sigmoidt_neuron(x=x, w=w, bias=bias)

plt.figure(figsize=(10, 6))
plt.plot(x[:, 0], y_tanh[0, :])
plt.title('Тангенциальный сигмоидный нейрон')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

y_complex = sigmoidt_complex_neuron(x=x, w=w, bias=bias, lymbd=lymbd)

plt.figure(figsize=(10, 6))
plt.plot(x[:, 0], y_complex[0, :])
plt.title(f'Сложный сигмоидный нейрон')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))

plt.plot(x[:, 0], y_linear[0, :], label='Линейный нейрон')
plt.plot(x[:, 0], y_sigmoid[0, :],  label='Сигмоидный нейрон')
plt.plot(x[:, 0], y_tanh[0, :], label='Тангенциальный сигмоидный нейрон')
plt.plot(x[:, 0], y_complex[0, :], label=f'Сложный сигмоидный нейрон')

plt.title('Сравнение выходов нейронов на одинаковом множестве входов')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()
plt.show()
```

# 1.3 Перечислить достоинства и недостатки активаций

| Функция активации | Достоинства | Недостатки |
|---|---|---|
| **Линейная** | • Простая в реализации и вычислении<br>  • Удобна для прогноза числового значения<br> | • Не добавляет нелинейность<br>• Не подходит для решения сложных нелинейных задач<br>• выход не ограничен диапазоном|
| **Сигмоидная** | • Плавная функция, выход находится в диапазоне от 0 до 1<br> • Удобна для интерпретации результата как степени принадлежности к классу.| • При больших по модулю входных значениях возникает насыщение<br>• В области насыщения изменение входа слабо влияет на выход |
| **Тангенциальная сигмоидная** | • Плавная функция, выход находится в диапазоне от -1 до 1 | • При больших по модулю входных значениях возникает насыщение<br>• В области насыщения изменение входа слабо влияет на выход |
| **Сложная сигмоидная** | • Позволяет управлять крутизной функции с помощью коэффициента λ | • При больших значениях λ быстро насыщается |

# 3.1. Сделать реализацию обучения по Дельта-правилу (правило Вудро-Хофа) для линейного нейрона

1. Задать начальное состояние
2. Взять пример j из набора примеров
3. Вычислить выход нейрона для этого примера
4. Вычислить коррекцию синаптических весов для всех входов xi

dwi = - n (y(Хj) - dj) xi

5. Вычислить новый уровень wi
wi = wi +  dwi

6. Перейти к новому примеру в п.2
7. Если просмотрели все примеры, то проверить условие конца (ошибка модели или число повторов):
    - условие выполнено - конец
    - условие не выполнено - начинаем просматривать примеры и изменять веса сначала (j=0) и к п.2

```python
def train_linear_delta(x, D, w, n, bias, max_epochs, eps):
    w = w.copy()

    y_before = linear_neuron(x=x, w=w, bias=0)

    error_history = []

    for epoch in range(max_epochs):
        for j in range(len(x)):
            x_j = x[j]
            d_j = D[j]

            y_j = linear_neuron(x=x_j, w=w, bias=0)

            error_j = y_j - d_j
            dw = -n * error_j * x_j

            w = w + dw

        y_all = linear_neuron(x=x, w=w, bias=0)

        errors = y_all - D

        mse = np.mean(errors ** 2)
        error_history.append(mse)

        if np.all(np.abs(errors) < eps):
            break

    y_after = y_all.copy()

    return w, errors, epoch + 1, y_before, y_after, error_history
```

# 3.2. Сделать реализацию обучения по Дельта-правилу для сигмоидного нейрона

```python
def train_sigmoid_delta(x, D, w, n, bias, max_epochs, eps):
    w = w.copy()

    y_before = sigmoidt_neuron(x=x, w=w, bias=0)

    error_history = []

    for epoch in range(max_epochs):
        for j in range(len(x)):
            x_j = x[j]
            d_j = D[j]

            y_j = sigmoidt_neuron(x=x_j, w=w, bias=0)

            error_j = y_j - d_j
            derivative = 0.5 * (1 - y_j ** 2)
            dw = -n * error_j * derivative * x_j

            w = w + dw

        y_all = sigmoidt_neuron(x=x, w=w, bias=0)

        errors = y_all - D

        mse = np.mean(errors ** 2)
        error_history.append(mse)

        if np.all(np.abs(errors) < eps):
            break

    y_after = y_all.copy()

    return w, errors, epoch + 1, y_before, y_after, error_history
```
